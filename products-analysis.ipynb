# Products Analysis

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv(r"data/products.csv")
df.head(4)
```

**Calculate the profit on each product after selling on list_price**

```python
df["profit"] = df["list_price"] - df["unit_cost"]
df.head(4)
```

**Find which cateogry has the most profit**

```python
fig = plt.figure(figsize=(14,8), dpi=100)
sns.barplot(data=df, x="category", y="profit", palette="rainbow", hue="category", estimator=np.mean)
```

**For each category find which sub-category had the most profit**

```python
fig = plt.figure(figsize=(14,14), dpi=100)
cats = df["category"].unique()
for index, cat in enumerate(cats):
    cat_df = df[df["category"]==cat]
    sub_fig = plt.subplot(3,2,index+1)
    plt.xlabel(cat)
    plt.ylabel("Profit")
    sns.barplot(data=cat_df, x="subcategory", y="profit", palette="rainbow", hue="subcategory", estimator=np.mean)
```
