# Order Return Analysis

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv(r"data/returns.csv")
df.head(4)
```

**Find by which reason most orders are returned**

```python
fig = plt.figure(figsize=(14,8), dpi=200)
sns.countplot(data=df, y="return_reason", palette="rainbow", hue="return_reason") # Most people return orders due to change of mind
```

**Find the reason of return that beared the most refund**

```python
fig = plt.figure(figsize=(14,8), dpi=100)
sns.barplot(data=df, y="return_reason", x="refund_amount", estimator=np.mean)
```
